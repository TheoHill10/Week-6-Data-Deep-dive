## Control Flow & Function

```python
x = 10

if x > 5:
    print(f'{x} is greater than 5')
```

```python
x = 2

if x > 5:
    print('x is greater than 5')
else: 
    print('x is 5 or smaller')
```

```python
x = 5

if x > 5:
    print('x is greater than 5')
elif x==5:
    print('x is equal to 5')
else: 
    print('x is smaller than x')
```

- conditional expression 
    - Syntax `value_if_true if condition else value_if_false`

```python
x = 2

result = 'x is greater than 5' if x > 5 else 'x is 5 or smaller'
print(result)
```

### for loop

- Syntax - `for num in numbers`

```python
numbers = [1, 2, 3 , 4, 5]

for num in numbers:
    add = num + 1
    print(add)
```

```python
for i in range(1, 20, 2): # range(start, stop, step)
    print(f'i is {i}')
```

```python
for i in range(10):
    if i % 2 == 0:
        print(f'{i} is even')
```

```python
total = 0

for r in rain:
    total += r
print(f'Total rainfall: {round(total, 1)}mm')
```

```python
for key, value in user.items():
    print(f'{key}: {value}')
```

```python
def greet(name):
    return f'Hello {name}'

message = greet('Theo')
print(message)
```
