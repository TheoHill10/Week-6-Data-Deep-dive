# Visulisations with pandas

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
```

```python
#line plotting with pandas

line_plot_data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Sales': [200, 220, 230, 210, 240, 250, 270, 260, 280, 300, 320, 330]
})

line_plot_data.plot(x='Month', y='Sales', kind='line', title='Sales by month')
```

```python
#line plotting with seaborn

seaborn_line_plot_data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Sales': [200, 220, 230, 210, 240, 250, 270, 260, 280, 300, 320, 330]
})

sns.lineplot(x='Month', y='Sales', data=seaborn_line_plot_data, color='indigo').set(title='Monthly Sales Trend')
```

```python
#line plotting in matplotlib

Month = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
Sales = [200, 220, 230, 210, 240, 250, 270, 260, 280, 300, 320, 330]


plt.figure(figsize=(8, 4))
plt.plot(Month, Sales, marker='o', color='violet')
plt.title('Sales per Month')
plt.xlabel('Month')
plt.ylabel('Sales')
```

```python
#bar plot with pandas

bar_plot_data = pd.DataFrame({
    'Speed': [0.1, 17.5, 40, 48, 52, 69, 88],
    'Lifespan': [2, 8, 70, 1.5, 25, 12, 28],
    'Index': ['snail', 'pig', 'elephant', 'rabbit', 'giraffe', 'coyote', 'horse']
})

bar_plot_data.plot(x='Index', y=['Lifespan', 'Speed'], kind='bar', title='Animals Lifespan vs Speed', color=['plum', 'rosybrown'])
```

```python
#bar plot in seaborn

seaborn_bar_plot_data = pd.DataFrame({
    'Flavour': ['Vanilla', 'Chocolate', 'Strawberry', 'Mint', 'Cookie Dough'],
    'Scoops_Sold': [120, 150, 90, 70, 130] 
})

sns.barplot(x='Flavour', y='Scoops_Sold', color= 'tan', data=seaborn_bar_plot_data).set(title='Ice cream Flavour vs Scoops Sold')
```

```python
# bar plot with matplot

genres = ['Fiction', 'Non-fiction', 'Mystery', 'Fantasy', 'Science', 'Romance', 'History']
books_owned = [50, 5, 37, 77, 8, 89, 5]

plt.figure(figsize=(8, 5))
plt.bar(genres, books_owned, color='seagreen')
plt.title('Books Owned by Genre')
plt.xlabel('Genre')
plt.xticks(rotation=45)
plt.ylabel('Books')
```

```python
# pie plot with matplotlib
groups = ['Fellowship', 'Villains', 'Allies', 'Others']
counts = [9, 7, 5, 3]

colours = ['gold', 'red', 'lightgreen', 'lightblue']
explode = [0.1, 0, 0, 0]

plt.pie(counts, labels=groups, autopct='%1.1f%%', startangle=90, colors=colours, explode=explode)
plt.title('Lord of the Rings Groups')
```

```python
#box and whisker plot with matplotlib

class_A = [55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 30] # low outlier

class_B = [50, 55, 58, 60, 63, 65, 67, 70, 73, 75, 90, 98] # high outlier

class_C = [65, 68, 70, 72, 74, 76, 78, 80, 82, 85, 88, 100] # extreme high outlier

plt.boxplot([class_A, class_B, class_C], tick_labels=['Class A', 'Class B', 'Class C'], showfliers=True, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.title('Exam Score Distribution by Class')
plt.ylabel('Score')
```
